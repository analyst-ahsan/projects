#### Preprocessing Insights EDA

Numerical Variables - Key Findings

Features Requiring Transformation (Highly Skewed):
1. **Tenure** → Log transform + binning (0-6mo, 6-12mo, 1-2yr, 2yr+)
2. **CouponUsed** → Log transform + create usage categories
3. **OrderCount** → Log transform + engagement categories (Critical: Low orders = high churn risk)
4. **DaySinceLastOrder** → Log transform + recency bins (CRITICAL RFM metric)

Features with Good Distribution:
- **OrderAmountHikeFromlastYear** → Needs only standardization
- **CashbackAmount** → Relatively clean, minimal preprocessing

Key Feature Engineering:
- Created recency categories (customers inactive 14+ days = high risk)
- Created engagement levels based on order count
- Created interaction features (cashback_per_order, avg_order_hike)
- Binary flags for quick segmentation (is_repeat_customer, active_last_week)

Outlier Handling:
- Capped WarehouseToHome at 95th percentile (extreme distances 120+ likely errors)
- Log transforms handle outliers in other variables naturally

```python
# ===== 1. Handle Skewness =====
# Log transforms for highly skewed variables
df['tenure_log'] = np.log1p(df['Tenure'])
df['warehouse_distance_log'] = np.log1p(df['WarehouseToHome'])
df['coupon_used_log'] = np.log1p(df['CouponUsed'])
df['order_count_log'] = np.log1p(df['OrderCount'])
df['days_since_last_order_log'] = np.log1p(df['DaySinceLastOrder'])

# ===== 2. Feature Engineering - Binning =====
# Tenure groups
df['tenure_group'] = pd.cut(df['Tenure'], 
                             bins=[-1, 6, 12, 24, 100], 
                             labels=['0-6mo', '6-12mo', '1-2yr', '2yr+'])

# Recency categories (CRITICAL for churn)
df['recency_category'] = pd.cut(df['DaySinceLastOrder'], 
                                 bins=[-1, 3, 7, 14, 30, 1000], 
                                 labels=['VeryRecent', 'Recent', 'Medium', 'Old', 'VeryOld'])

# Customer engagement
df['customer_engagement'] = pd.cut(df['OrderCount'], 
                                    bins=[0, 2, 4, 8, 100], 
                                    labels=['Low', 'Medium', 'High', 'VeryHigh'])

# ===== 3. Binary Flags =====
df['is_repeat_customer'] = (df['OrderCount'] > 1).astype(int)
df['active_last_week'] = (df['DaySinceLastOrder'] <= 7).astype(int)
df['has_used_coupon'] = (df['CouponUsed'] > 0).astype(int)

# ===== 4. Interaction Features =====
df['cashback_per_order'] = df['CashbackAmount'] / (df['OrderCount'] + 1)
#df['avg_order_hike'] = df['OrderAmountHikeFromlastYear'] / (df['OrderCount'] + 1)

# ===== 5. Cap Outliers (optional) =====
df['warehouse_distance_capped'] = df['WarehouseToHome'].clip(
    upper=df['WarehouseToHome'].quantile(0.95)
)

print("Preprocessing complete for numerical vars!")
```

Categorical Variables - Key Findings

Strong Predictors (Definitely Keep):
- **Complain**: 28.5% have complained → likely strong churn indicator
- **SatisfactionScore**: Ordinal 1-5 → dissatisfied customers (≤2) will churn more
- **PreferredPaymentMode**: Payment behavior often correlates with churn
- **CityTier**: Tier 1 vs others might show different churn patterns

Potential Weak Predictors (Check Bivariate):
- **Gender**: 60-40 split, but might have similar churn rates
- Check if churn rates are similar across genders before keeping

Encoding Decisions:
- **One-hot encode**: Low cardinality vars (≤3 categories)
- **Target encode**: PreferredPaymentMode (7 categories)
- **Keep numeric**: Ordinal vars (SatisfactionScore, CityTier, DeviceCount)
- **Create flags**: is_tier1_city, is_dissatisfied, has_complained

Feature Engineering from Categoricals:
- Create interaction: complain × satisfaction_score
- Create device_to_address_ratio: NumberOfDeviceRegistered / NumberOfAddress
- Payment preference + Order category interaction

```python
# ===== 1. Clean Duplicates =====
# Merge similar payment methods
df['PreferredPaymentMode'] = df['PreferredPaymentMode'].replace({
    'Cash on Delivery': 'COD',
    'Credit Card': 'CC',
})

# ===== 2. Binary Variables (Keep as 0/1) =====
df['gender_encoded'] = df['Gender'].map({'Male': 0, 'Female': 1})
df['has_complained'] = df['Complain']  # Already 0/1

# ===== 3. Ordinal Variables (Preserve Order) =====
# Already numeric, keep as-is
ordinal_vars = ['CityTier', 'SatisfactionScore', 'NumberOfDeviceRegistered', 'HourSpendOnApp']

# Create additional binary flags
df['is_tier1_city'] = (df['CityTier'] == 1).astype(int)
df['is_dissatisfied'] = (df['SatisfactionScore'] <= 2).astype(int)
df['low_engagement'] = (df['HourSpendOnApp'] <= 2).astype(int)

# ===== 4. Low Cardinality - One-Hot Encoding =====
# These have 2-3 categories
onehot_cols = ['PreferredLoginDevice', 'MaritalStatus']
df_encoded = pd.get_dummies(df, columns=onehot_cols, drop_first=True)  # drop_first to avoid multicollinearity

# ===== 5. Medium Cardinality - Group + One-Hot =====
# PreferedOrderCat: Group smallest categories
def group_order_cat(x):
    if x in ['Laptop & Accessory', 'Mobile Phone', 'Fashion', 'Mobile']:
        return x
    return 'Other'

df_encoded['order_cat_grouped'] = df['PreferedOrderCat'].apply(group_order_cat)
df_encoded = pd.get_dummies(df_encoded, columns=['order_cat_grouped'], drop_first=True)

# ===== 6. High Cardinality - Target Encoding =====
# PreferredPaymentMode: Use target encoding
te = TargetEncoder(smooth="auto")
df_encoded['Payment_Encoded'] = te.fit_transform(df_encoded[['PreferredPaymentMode']], df_encoded['Churn'])

print("Preprocessing complete for categ vars!")
print(f"Encoded features: {len(df_encoded.columns)}")
```

#### Modelling

```python
y = df_encoded['Churn']
X = df_encoded.drop(columns=['Churn'])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2, 
    stratify=y,  
    random_state=42
)

print(f"Train churn rate: {y_train.mean():.2%}")
print(f"Test churn rate: {y_test.mean():.2%}")
```
